```python

import pandas as pd
```

```python

data = {
    'Student_ID': [1, 2, 3, 4, 5],
    'Name': ['Anu', 'Amal', 'Sara', 'John', 'Riya'],
    'Maths': [85, 70, 90, 65, 80],
    'Science': [80, 75, 85, 70, 90],
    'English': [90, 80, 88, 75, 85]
}

df = pd.DataFrame(data)

df
```

```python

# Display first 5 rows
df.head()
```

```python

# Check the number of rows and columns
print(df.shape)

# Display column names
print(df.columns)
```

```python

df.info()
```

```python

# Calculate total marks
df['Total'] = df['Maths'] + df['Science'] + df['English']

# Calculate average marks
df['Average'] = df['Total'] / 3

# Display the result
df
```

```python

print("Highest Average:", df['Average'].max())

print("Lowest Average:", df['Average'].min())

print("Class Average:", df['Average'].mean())
```

```python

top_student = df.loc[df['Average'].idxmax()]

print("Top Student:", top_student['Name'])
print("Average Marks:", top_student['Average'])
```

```python

low_student = df.loc[df['Average'].idxmin()]

print("Lowest Student:", low_student['Name'])
print("Average Marks:", low_student['Average'])
```

```python

import matplotlib.pyplot as plt
```

```python

subjects = ['Maths', 'Science', 'English']

averages = [
    df['Maths'].mean(),
    df['Science'].mean(),
    df['English'].mean()
]

plt.bar(subjects, averages)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.bar(df['Name'], df['Average'])

plt.title("Student Performance Comparison")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")

plt.show()
```

```python

plt.hist(df['Average'], bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")

plt.show()
```
